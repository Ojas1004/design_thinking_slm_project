### Cell 1: Load CSV using built-in csv module
Define the input CSV filename, load it, and display the first 3 rows to confirm it loaded correctly.

```python
import csv

# Define the input CSV path (change this variable if your filename changes)
input_csv_path = '/content/energydata_complete.csv'

csv_data = []
with open(input_csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.reader(f)
    # Extract header and rows
    header = next(reader)
    for row in reader:
        csv_data.append(row)

# Print the header and the first 3 rows to confirm
print("Header:", header)
print("First 3 Rows:")
for row in csv_data[:3]:
    print(row)
```

### Cell 2: Convert CSV to JSON format
Convert the parsed CSV list into a standard Python list of dictionaries (JSON format) and preview the JSON data.

```python
import json

json_data = []
for row in csv_data:
    # Map the header to row values
    record = dict(zip(header, row))

    # Restructure into a standard prompt-completion pair suitable for an SLM
    prompt = f"Predict the appliances energy use (Wh) for date {record.get('date')} given the following conditions: Temperature in kitchen is {record.get('T1')}°C, Humidity in kitchen is {record.get('RH_1')}%. Outside Temperature is {record.get('T_out')}°C."
    completion = f"The predicted appliances energy use is {record.get('Appliances')} Wh."

    json_data.append({
        "prompt": prompt,
        "completion": completion
    })

# Display a preview of the parsed JSON data
print(json.dumps(json_data[:3], indent=2))
```

### Cell 3: Convert JSON data to JSONL format
Export the list of dictionaries line-by-line into a JSONL file named `output.jsonl`.

```python
output_jsonl_path = 'output.jsonl'

with open(output_jsonl_path, mode='w', encoding='utf-8') as f:
    for item in json_data:
        f.write(json.dumps(item) + '\n')

print(f"Success! JSONL file successfully created at: {output_jsonl_path}")
```

```python
# Read back the generated output.jsonl file line-by-line
print("Verifying generated JSONL contents:")
with open(output_jsonl_path, mode='r', encoding='utf-8') as f:
    for i in range(5):
        line = f.readline()
        if not line:
            break
        # Parse each line to verify it is a valid independent JSON object
        record = json.loads(line.strip())
        print(f"Record {i+1}:", json.dumps(record, indent=2))
        print("-" * 40)
```
